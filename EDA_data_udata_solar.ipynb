# Exploratory Data Analysis for data_udata_solar.csv

This notebook performs EDA on the real solar dataset.

```python
# Import libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
# Load dataset
df = pd.read_csv('data_udata_solar.csv')
df.head()
```

```python
# Dataset info and statistics
display(df.info())
display(df.describe())
```

```python
# Check missing values
df.isnull().sum()
```

```python
# Correlation heatmap
plt.figure(figsize=(10, 6))
sns.heatmap(df.corr(), annot=True, cmap='coolwarm')
plt.title('Correlation Heatmap')
plt.show()
```

```python
# Distribution of target variable (if present)
if 'solar_power' in df.columns:
    plt.figure(figsize=(8, 4))
    sns.histplot(df['solar_power'], bins=50, kde=True)
    plt.title('Solar Power Distribution')
    plt.show()
```
